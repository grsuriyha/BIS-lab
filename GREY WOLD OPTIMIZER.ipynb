```python

import random
import math

def fitness(x):
    return x*x - 10*math.cos(2*math.pi*x) + 10

wolves = [random.uniform(-5, 5) for _ in range(5)]

for _ in range(20):
    wolves.sort(key=fitness)

    alpha = wolves[0]
    beta = wolves[1]
    delta = wolves[2]

    for i in range(3, len(wolves)):
        wolves[i] = (alpha + beta + delta) / 3

print("Best Solution:", alpha)
print("Best Fitness:", fitness(alpha))
```
